```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)

df = pd.read_csv('titanic_raw.csv')

print("="*70)
print("SHAPE")
print("="*70)
print(df.shape)

print("\n" + "="*70)
print("DTYPES")
print("="*70)
print(df.dtypes)

print("\n" + "="*70)
print("HEAD")
print("="*70)
print(df.head(10))

print("\n" + "="*70)
print("DESCRIBE (numeric)")
print("="*70)
print(df.describe())

print("\n" + "="*70)
print("DESCRIBE (categorical)")
print("="*70)
print(df.describe(include=['object']))

print("\n" + "="*70)
print("MISSING VALUES")
print("="*70)
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_table = pd.DataFrame({'missing_count': missing, 'missing_pct': missing_pct})
missing_table = missing_table[missing_table.missing_count > 0].sort_values('missing_count', ascending=False)
print(missing_table)

print("\n" + "="*70)
print("DUPLICATE ROWS")
print("="*70)
print("Exact duplicate rows:", df.duplicated().sum())
print("Duplicate PassengerId:", df['PassengerId'].duplicated().sum())

print("\n" + "="*70)
print("UNIQUE VALUES - key categoricals")
print("="*70)
for col in ['Survived','Pclass','Sex','SibSp','Parch','Embarked']:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))

print("\n" + "="*70)
print("RANGE CHECKS")
print("="*70)
print("Age min/max:", df['Age'].min(), df['Age'].max())
print("Fare min/max:", df['Fare'].min(), df['Fare'].max())
print("Fare == 0 count:", (df['Fare']==0).sum())
print("Age fractional values (children under 1yr encoded as fractions):")
print(df.loc[df['Age'] % 1 != 0, 'Age'].dropna().unique())
```
